## 1. Daily and weekly steps

```python
daily_steps = {"Sun" : 6000, "Mon" : 7500, "Tue" : 8000, "Wed" : 9000, "Thu" : 10000, "Fri" : 11000, "Sat" : 12000}
```

```python
def disp_daily_steps(steps):
    for day, step_count in steps.items():
        print(f"{day}: {step_count} steps")

disp_daily_steps(daily_steps)
```

```python
def disp_weekly_steps(steps):
    total_steps = sum(steps.values())
    print(f"Total steps for the week: {total_steps} steps")

disp_weekly_steps(daily_steps)
```

## 2. Heart Rate Analysis

```python
import random
```

```python
heart_rates = []
for i in range(14):
    heart_rate = random.randint(60, 100)
    heart_rates.append(heart_rate)
```

```python
def disp_heart_rates(rates):
    return ("Avg Heart Rate:", sum(rates)/len(rates))

disp_heart_rates(heart_rates)
```

## 3. Calories & Activity Duration

```python
weekly_activities = [
    {"day": "Saturday",  "calories_burned": 280, "duration": 35},
    {"day": "Sunday",    "calories_burned": 190, "duration": 25},
    {"day": "Monday",    "calories_burned": 410, "duration": 50},
    {"day": "Tuesday",   "calories_burned": 320, "duration": 40},
    {"day": "Wednesday", "calories_burned": 150, "duration": 20},
    {"day": "Thursday",  "calories_burned": 460, "duration": 55},
    {"day": "Friday",    "calories_burned": 380, "duration": 45}
]
```

```python
def disp_weekly_calories_and_duration(weekly_activities):
    total_calories = sum(activity["calories_burned"] for activity in weekly_activities)
    total_duration = sum(activity["duration"] for activity in weekly_activities)
    print(f"Total calories burned for the week: {total_calories} kcal")
    print(f"Total duration of workouts for the week: {total_duration} minutes")

disp_weekly_calories_and_duration(weekly_activities)
```

## 4. Daily Activity Summary

```python
user_daily_activity = {
    "day": "Wednesday",
    "steps": 11200,
    "calories": 460,
    "duration": 55,
    "heart_rate": 84
}
```

```python
def user_activity_summary(activity):
    print(f"Activity Summary for {activity['day']}:")
    print(f"Steps: {activity['steps']}")
    print(f"Calories Burned: {activity['calories']} kcal")
    print(f"Duration: {activity['duration']} minutes")
    print(f"Average Heart Rate: {activity['heart_rate']} bpm")

user_activity_summary(user_daily_activity)
```
